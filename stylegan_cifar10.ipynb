# StyleGAN for CIFAR-10 in PyTorch

Production-grade PyTorch implementation of StyleGAN (Karras et al.) trained on CIFAR-10 ($32 \times 32$ RGB resolution).

### Architectural Innovations
- **Mapping Network (Z -> W)**: 4-layer MLP mapping $z \in \mathbb{R}^{128}$ to $w \in \mathbb{R}^{128}$.
- **Adaptive Instance Normalization (AdaIN)**: Style vector $w$ dynamically modulates channel scales and shifts.
- **Learned Constant Input Base**: Starts synthesis from a learned constant tensor $C \in \mathbb{R}^{1 \times 128 \times 4 \times 4}$ initialized to ones.
- **Stochastic Noise Injection**: Per-pixel Gaussian noise added with learned scaling factors after each convolution.
- **Minibatch Standard Deviation**: Prevents mode collapse in the discriminator.
- **Label Smoothing**: Real targets smoothed to 0.9 for training stability.
